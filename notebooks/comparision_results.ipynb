```python
"""
05_comparison.ipynb
Combines results from all trained models and writes a final comparison.
"""

import json
import sys
import os

sys.path.append("../src")
from utils import save_results_json
```

```python
# Load individual result files saved during Day 3 and Day 4

with open("../models/tfidf_results.json", "r") as f:
    tfidf_results = json.load(f)  # this should be a list: [logreg, xgb, bayes, svm]

with open("../models/distilbert_model.json", "r") as f:
    distilbert_results = json.load(f)
```

```python
# Combine everything into one list of results
all_results = tfidf_results + [distilbert_results]

# Save combined comparison file
save_results_json(all_results, "../models/comparison_results.json")
```

```python
# Build a clean comparison table for display / README
import pandas as pd

comparison_df = pd.DataFrame([
    {
        "model": r.get("model_name", r.get("model", "DistilBERT")),
        "accuracy": r.get("accuracy", r.get("eval_accuracy")),
        "precision": r.get("precision", r.get("eval_precision")),
        "recall": r.get("recall", r.get("eval_recall")),
        "f1_score": r.get("f1_score", r.get("eval_f1")),
    }
    for r in all_results
])

comparison_df = comparison_df.sort_values("f1_score", ascending=False).reset_index(drop=True)
comparison_df
```

```python
# Save the clean comparison table as CSV too, useful for README/portfolio
comparison_df.to_csv("../models/comparison_table.csv", index=False)
```
